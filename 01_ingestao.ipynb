```python
import os
import zipfile
from dotenv import load_dotenv

# 1. Carrega as variáveis do arquivo .env
load_dotenv(override=True)

# 2. Configura as credenciais para o Kaggle usar
kaggle_user = os.getenv("KAGGLE_USERNAME")
kaggle_key = os.getenv("KAGGLE_KEY")

# Trava de segurança: Se não achar as senhas, para tudo e avisa
if not kaggle_user or not kaggle_key:
    raise ValueError("ERRO CRÍTICO: Credenciais do Kaggle não encontradas no arquivo .env!")

# Passa as credenciais para as variáveis de ambiente que o comando !kaggle exige
os.environ['KAGGLE_USERNAME'] = kaggle_user
os.environ['KAGGLE_KEY'] = kaggle_key

# 3. Definição de Caminhos
dataset_name = "olistbr/brazilian-ecommerce"
path_destino = "/Volumes/olist_portfolio/bronze/raw_data"

print(f"--- Iniciando Ingestão de: {dataset_name} ---")

# 4. Download (Kaggle CLI)
print("1. Baixando arquivo ZIP do Kaggle...")
# O comando !kaggle usa as variáveis de ambiente que configuramos acima
!kaggle datasets download -d {dataset_name} -p {path_destino}

# 5. Descompactação
print("2. Descompactando arquivos...")
zip_file = f"{path_destino}/brazilian-ecommerce.zip"

if os.path.exists(zip_file):
    with zipfile.ZipFile(zip_file, 'r') as z:
        z.extractall(path_destino)
    print("   -> Descompactação concluída com sucesso!")
else:
    print(f"   -> ERRO: O arquivo {zip_file} não foi encontrado. O download falhou?")

print(f"\n✅ Processo Finalizado! Arquivos salvos em: {path_destino}")

# Mostra os arquivos para confirmar
display(dbutils.fs.ls(path_destino))
```
